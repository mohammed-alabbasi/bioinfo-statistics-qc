# Example 4 (R half): Testing the Replichore Asymmetry — Is the Split Significant?


Read the skew table this Example's Python half just wrote, run a Wilcoxon rank-sum test comparing the replichores , and produce a publication-style boxplot with the p-value annotated. **Use the R kernel.**



```r
# Test whether GC skew differs significantly between the two replichores, and plot it
library(ggplot2)

dir.create("results", showWarnings = FALSE)  # this Example creates the folder it needs 

skew_data <- read.csv("results/example4_skew_values.csv")  # written by this Example's own Python half

skew_data$replichore <- factor(skew_data$replichore, levels = c("replichore_1", "replichore_2"))

test_result <- wilcox.test(skew ~ replichore, data = skew_data)
cat("Wilcoxon rank-sum test\n")
cat(sprintf("W = %.1f, p-value = %.2e\n", test_result$statistic, test_result$p.value))

p <- ggplot(skew_data, aes(x = replichore, y = skew, fill = replichore)) +
  geom_boxplot(alpha = 0.7, outlier.size = 0.8) +
  geom_hline(yintercept = 0, linetype = "dashed", color = "gray40") +
  labs(
    title = "GC skew by replichore (E. coli K-12 MG1655)",
    subtitle = sprintf("Wilcoxon rank-sum test: p = %.2e", test_result$p.value),
    x = "Replichore", y = "GC skew (G-C)/(G+C)"
  ) +
  theme_minimal() +
  theme(legend.position = "none")

ggsave("results/example4_skew_boxplot.png", p, width = 6, height = 4.5, dpi = 150)
cat("Saved: results/example4_skew_boxplot.png\n")
```
